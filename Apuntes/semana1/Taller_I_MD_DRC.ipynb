## Promedio

$$\bar{x} = \frac{1}{n} \sum_{i} x_i$$

```python
import statistics as st
data = [2, 2, 2, 2, 3, 4, 4, 6, 7, 8, 8, 9, 10, 11, 35]
x_mean = st.mean(data)
print("Promedio:", x_mean)
```

```python
import numpy as np
x_mean = (1/len(data))*np.sum(data)
print("Promedio:", x_mean)
```

## Varianza

$$\sigma_{x}^{2} = \frac{1}{n-1} \sum_{i} (x_i - \bar{x})^2$$

```python
# Cálculo de la varianza
x_var = st.variance(data)
print("Varianza:", x_var)
```

```python
# Cálculo de la varianza
x_var = (1/(len(data)-1))*np.sum((np.array(data) - x_mean)**2)
print("Varianza:", x_var)
```

## Desviacion estandar

$$\sigma_{x} = \sqrt{\sigma_[{x}^{2}}$$

```python
# Cálculo de la desviación estándar
x_std = st.stdev(data)
print("Desviación Estándar:", x_std)
```

```python
# Cálculo de la desviación estándar
x_std = np.sqrt(x_var)
print("Desviación Estándar:", x_std)
```

## Covarianza

$$\sigma_{xy} = \frac{1}{n-1} \sum_{i} (x_i - \bar{x})(y_i - \bar{y})$$

```python
# Cálculo de la covarianza
data1 = [2, 2, 2, 2, 3, 4, 4, 6, 7, 8, 8, 9, 10, 11, 35]
data2 = [2, 4, 5, 6, 3, 4, 4, 6, 7, 8, 8, 9, 10, 11, 20]
x_cov = np.cov(data1,data2)
print("Covarianza:", x_cov)
```

```python
# Cálculo de la covarianza
x_cov = (1/(len(data)-1))*np.sum((data1 - np.mean(data1))*(data2 - np.mean(data1)))
print("Covarianza:", x_cov)
```

## Correlacion

$$\rho_{xy} = \frac{\sigma_{xy}}{\sigma_{x}\sigma_{y}}$$

```python
# Cálculo de la correlacion
x_cor = np.cov(data1, data2, ddof=1)[0,1] / (np.std(data1, ddof=1) * np.std(data2, ddof=1))
print("Correlacion:", x_cor)

```

## Error estandar

$$SE(x) = \frac{\sigma_{x}}{\sqrt{n}}$$





```python
# Cálculo del error estandar
x_es = np.std(data1)/(np.sqrt(len(data1)))
print("Error Estandar:", x_es)
```

## Mediana

Para valores impares

$$x_m = x \left [ \frac{n+1}{2}\right ]$$

Para valores pares

$$x_m = \frac{1}{2} \left (  x \left [\frac{n}{2}\right ] + x \left [\frac{n}{2} + 1 \right ]\right )$$


```python
# Cálculo de la mediana
x_med = st.median(data1)
print("Mediana:", x_med)
```

## Moda

La mode corresponde al valor que mas se repite en nuestro datos

```python
# Cálculo de la moda
data1 = [2, 2, 2, 2, 3, 4, 4, 4, 4, 8, 8, 9, 10, 11, 35]
x_mod = st.mode(data1)
print("Moda:", x_mod)
```

El rango se define como la diferencia entre el valor maximo y el valor mínimo de la variable analizada.

## Distribución normal de una variable

La distribución normal es un modelo teórico capaz de aproximar satisfactoriamente el valor de una variable aleatoria a una situación ideal.
Tambien esta se puede presentar en formato de probabilidad de la siguiente manera,

$$p(x)=\frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{(u-\mu)^{2}}{2\sigma^{2}}}$$

```python
# histogramas de una variable en este caso es una distribucion normal con media 0 y desviacion estandar 1.
# el histograma de la izquierda es el histograma sin normalizar y el histograma de la derecha es el histograma normalizado (probabilidad)
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=42)
a = np.hstack((rng.normal(loc=0, scale=1, size=10000)))

fig, ax = plt.subplots(1,2,figsize=(15,6))
ax[0].hist(a, bins='auto')
ax[0].set_title("Histograma con agrupamiento automático")
ax[0].set_xlabel("valor")
ax[0].set_ylabel("frecuencia")

ax[1].hist(a, bins='auto', density=True, color='g')
ax[1].set_title("Histograma con agrupamiento automático")
ax[1].set_xlabel("valor")
ax[1].set_ylabel("probabilidad")
plt.tight_layout()
plt.show()
```
