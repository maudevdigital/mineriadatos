##  Error Absoluto Máximo (M)

El **error absoluto máximo**, denotado como **M**, mide la **mayor diferencia absoluta** entre los valores verdaderos \(y_i\) y los valores predicho
(ŷᵢ)

$$
M = \max_{i} \left| y_i - \hat{y}_i \right|
$$

---

###  Donde:

- `y_i`: valor verdadero (real)  
- `ŷ_i`: valor predicho (estimado por el modelo)  
- `| · |`: valor absoluto  
- `max`: operador que selecciona el valor máximo entre todos los errores

---

###  Ejemplo:

| Valor real (y_i) | Predicho (ŷ_i) | Error absoluto |
|------------------|----------------|----------------|
| 1                | 1              | 0              |
| 2                | 2              | 0              |
| 3                | 3              | 0              |
| 4                | 4              | 0              |
| 5                | -5             | 10             |

$$
M = \max \{0, 0, 0, 0, 10\} = 10
$$





```python
from sklearn.metrics import max_error

y_verdadero = [1, 2, 3, 4, 5]
y_predicho = [1, 2, 3, 4, -5]

resultado = max_error(y_verdadero, y_predicho)
entero = int(resultado)

print(entero)
```

##  Error Absoluto Medio (MAE)

El **error absoluto medio**, conocido como **MAE** (*Mean Absolute Error*), mide el **promedio de las diferencias absolutas** entre los valores verdaderos (yᵢ) y los valores predichos (ŷᵢ):

$$
MAE = \frac{1}{n} \sum_{i=1}^{n} \left| y_i - \hat{y}_i \right|
$$

---

###  Donde:

- **yᵢ**: valor verdadero (real)  
- **ŷᵢ**: valor predicho (estimado por el modelo)  
- **n**: tamaño de la muestra (número total de observaciones)  
- **| · |**: valor absoluto

---

###  Ejemplo:

| Valor real (yᵢ) | Predicho (ŷᵢ) | Error absoluto |
|------------------|----------------|----------------|
| 1                | 1              | 0              |
| 2                | 2              | 0              |
| 3                | 3              | 0              |
| 4                | 4              | 0              |
| 5                | -5             | 10             |

$$
MAE = \frac{0 + 0 + 0 + 0 + 10}{5} = 2
$$


```python
from sklearn.metrics import mean_absolute_error

y_verdadero = [1, 2, 3, 4, 5]
y_predicho = [1, 2, 3, 4, -5]
mean_absolute_error(y_verdadero, y_predicho)
```

##  Error Cuadrático Medio (MSE)

El **error cuadrático medio**, conocido como **MSE** (*Mean Squared Error*), mide el **promedio de los errores al cuadrado** entre los valores verdaderos (yᵢ) y los valores predichos (ŷᵢ):

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} \left( y_i - \hat{y}_i \right)^2
$$

---

###  Donde:

- **yᵢ**: valor verdadero (real)  
- **ŷᵢ**: valor predicho (estimado por el modelo)  
- **n**: tamaño de la muestra (número total de observaciones)  
- **( · )²**: error al cuadrado

---

###  Ejemplo:

| Valor real (yᵢ) | Predicho (ŷᵢ) | Error (yᵢ - ŷᵢ) | Error² |
|------------------|----------------|------------------|--------|
| 1                | 1              | 0                | 0      |
| 2                | 2              | 0                | 0      |
| 3                | 3              | 0                | 0      |
| 4                | 4              | 0                | 0      |
| 5                | -5             | 10               | 100    |

$$
MSE = \frac{0 + 0 + 0 + 0 + 100}{5} = 20
$$


```python
from sklearn.metrics import mean_squared_error

y_verdadero = [1, 2, 3, 4, 5]
y_predicho = [1, 2, 3, 4, -5]
mean_squared_error(y_verdadero, y_predicho)
```

##  Suma de los Cuadrados de los Residuos (RSS)

La **Suma de los Cuadrados de los Residuos**, conocida como **RSS** (*Residual Sum of Squares*), mide el total de la variación no explicada por el modelo. Se calcula sumando los cuadrados de las diferencias entre los valores verdaderos (yᵢ) y los valores predichos (ŷᵢ):

$$
RSS = \sum_{i=1}^{n} \left( y_i - \hat{y}_i \right)^2
$$

---

###  Donde:

- **yᵢ**: valor verdadero (real)  
- **ŷᵢ**: valor predicho (estimado por el modelo)  
- **n**: tamaño de la muestra (número total de observaciones)  
- **( · )²**: error al cuadrado

---

### Ejemplo:

| Valor real (yᵢ) | Predicho (ŷᵢ) | Error (yᵢ - ŷᵢ) | Error² |
|------------------|----------------|------------------|--------|
| 1                | 1              | 0                | 0      |
| 2                | 2              | 0                | 0      |
| 3                | 3              | 0                | 0      |
| 4                | 4              | 0                | 0      |
| 5                | -5             | 10               | 100    |

$$
RSS = 0 + 0 + 0 + 0 + 100 = 100
$$


```python
from sklearn.metrics import mean_squared_error

y_verdadero = [1, 2, 3, 5, 5]
y_predicho = [1, 2, 3, 4, -5]
mean_squared_error(y_verdadero, y_predicho)*len(y_predicho)
```

##  Raíz Cuadrada del Error Cuadrático Medio (RMSE)

La **raíz cuadrada del error cuadrático medio**, conocida como **RMSE** (*Root Mean Squared Error*), es una medida que indica el **promedio de la magnitud de los errores**, penalizando más fuertemente los errores grandes. Es simplemente la raíz cuadrada del MSE:


$$
\text{RMSE} = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 }
$$

---

###  Donde:

- **yᵢ**: valor verdadero (real)  
- **ŷᵢ**: valor predicho (estimado por el modelo)  
- **n**: tamaño de la muestra  
- **√ ·**: raíz cuadrada

---

###  Ejemplo:

| Valor real (yᵢ) | Predicho (ŷᵢ) | Error (yᵢ - ŷᵢ) | Error² |
|------------------|----------------|------------------|--------|
| 1                | 1              | 0                | 0      |
| 2                | 2              | 0                | 0      |
| 3                | 3              | 0                | 0      |
| 4                | 4              | 0                | 0      |
| 5                | -5             | 10               | 100    |

$$
RMSE = \sqrt{ \frac{0 + 0 + 0 + 0 + 100}{5} } = \sqrt{20} \approx 4.47
$$


```python
from sklearn.metrics import mean_squared_error
import numpy as np

y_verdadero = [1, 2, 3, 4, 5]
y_predicho = [1, 2, 3, 4, -5]

mse = mean_squared_error(y_verdadero, y_predicho)
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

##  Coeficiente de Determinación (\( R^2 \))

---



El **coeficiente de determinación**, denotado como \( R^2 \), mide la proporción de la varianza en la variable dependiente que puede ser explicada por el modelo. Es una métrica de **bondad de ajuste** del modelo.


$$
R^2 = 1 - \frac{ \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 }{ \sum_{i=1}^{n} (y_i - \bar{y})^2 }
$$

---

###  Donde:

$y_i$: valor verdadero (real)

$\hat{y}_i$: valor predicho por el modelo

$\bar{y}$: promedio de los valores verdaderos

$n$: número total de observaciones

---

###  Interpretación:

- \( R^2 = 1 \): el modelo explica el **100%** de la variabilidad de los datos.
- \( R^2 = 0 \): el modelo no explica **nada** de la variabilidad (igual a predecir con la media).
- \( R^2 < 0 \): el modelo es **peor** que predecir con la media (muy mal ajuste).

> Se utiliza principalmente para **evaluar modelos de regresión** y comparar su capacidad explicativa.


```python
from sklearn.metrics import r2_score

y_verdadero = [1, 2, 3, 4, 5]
y_predicho = [1, 2, 3, 4, 2]
r2_score(y_verdadero, y_predicho)
```
